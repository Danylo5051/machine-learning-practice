### Завдання 25
Дано два вектори:


````python
# Код студента
a = np.array([1, 2])
b = np.array([2, 1])

cos_theta = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
print("Косинус кута:", cos_theta)

angle_rad = np.arccos(cos_theta)
angle_deg = np.degrees(angle_rad)
print("Кут у градусах:", angle_deg)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array([[10, 20, 30],
              [20, 30, 40],
              [30, 40, 50]])

mu = np.mean(A, axis=0)
sigma = np.std(A, axis=0)

Z = (A - mu) / sigma
print("Нормалізована матриця (Z-оцінка):\n", Z)
````

```text
sales = [[120, 150, 170, 160],
         [80,  90,  100, 110],
         [200, 180, 210, 230]]
```

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;
2. загальний обсяг продажів у кожному місяці;
3. товар із найбільшими сумарними продажами;

Обчисліть **евклідову відстань** між:

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента
A = np.array([[ 1,  2,  1],
              [ 2,  1, -1],
              [ 3, -1,  2]])

b = np.array([9, 8, 3])
x = np.linalg.solve(A, b)
print("1. Розв'язок системи (x, y, z):", x)

A_times_x = A @ x
print("2. Результат матричного множення A @ x:", A_times_x)

is_correct = np.allclose(A_times_x, b)
print("3. збігається результат з вектором b", is_correct)
````
